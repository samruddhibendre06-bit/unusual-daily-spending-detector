```python
import pandas as pd

df = pd.read_csv("Daily Household Transactions.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print(df.shape)
print(df.columns.tolist())
display(df.head())
print(df.info())
```

```python
# Check dataset size
print("Rows and columns:", df.shape)

# Check income and expense counts
print("\nIncome/Expense counts:")
print(df["Income/Expense"].value_counts())

# Check currencies
print("\nCurrencies:")
print(df["Currency"].value_counts())

# Check transaction categories
print("\nCategories:")
print(df["Category"].value_counts())

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Keep an unchanged copy of the original data
df_clean = df.copy()

# Remove duplicate transactions
df_clean = df_clean.drop_duplicates().copy()

# Convert Date into proper date format
df_clean["Date"] = pd.to_datetime(
    df_clean["Date"], dayfirst=True, errors="coerce"
)

# Remove rows with invalid dates or amounts
df_clean = df_clean.dropna(subset=["Date", "Amount"])

# Remove transactions with zero or negative amounts
df_clean = df_clean[df_clean["Amount"] > 0].copy()

# Fill missing categorical information
df_clean["Category"] = df_clean["Category"].fillna("Unknown")
df_clean["Mode"] = df_clean["Mode"].fillna("Unknown")
df_clean["Subcategory"] = df_clean["Subcategory"].fillna("Unknown")
df_clean["Note"] = df_clean["Note"].fillna("No note")

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))

display(df_clean.head())
```

```python
# Check income/expense types in the cleaned dataset
print(df_clean["Income/Expense"].value_counts())

# Check currencies
print("\nCurrency counts:")
print(df_clean["Currency"].value_counts())

# Check amount statistics
print("\nAmount summary:")
print(df_clean["Amount"].describe())
```

```python
# Keep only expense transactions in Indian rupees
expenses = df_clean[
    (df_clean["Income/Expense"].str.strip().str.lower() == "expense") &
    (df_clean["Currency"].str.strip().str.upper() == "INR")
].copy()

print("Expense transactions available:", len(expenses))
display(expenses.head())
```

```python
# Create a separate dataset for machine learning
ml_data = expenses.copy()

# Extract date-based features
ml_data["DayOfWeek"] = ml_data["Date"].dt.dayofweek
ml_data["Month"] = ml_data["Date"].dt.month
ml_data["Hour"] = ml_data["Date"].dt.hour
ml_data["IsWeekend"] = (ml_data["DayOfWeek"] >= 5).astype(int)

# Log transformation reduces the effect of very large amounts
ml_data["LogAmount"] = np.log1p(ml_data["Amount"])

# Compare each expense with the typical amount in its category
category_median = ml_data.groupby("Category")["Amount"].transform("median")

ml_data["CategoryAmountRatio"] = (
    ml_data["Amount"] / category_median.replace(0, np.nan)
).fillna(1)

# Convert categories into numerical features
category_features = pd.get_dummies(
    ml_data["Category"],
    prefix="Category",
    dtype=int
)

mode_features = pd.get_dummies(
    ml_data["Mode"],
    prefix="Mode",
    dtype=int
)

# Combine numerical and categorical features
features = pd.concat(
    [
        ml_data[
            [
                "LogAmount",
                "DayOfWeek",
                "Month",
                "Hour",
                "IsWeekend",
                "CategoryAmountRatio"
            ]
        ],
        category_features,
        mode_features
    ],
    axis=1
)

features = features.replace([np.inf, -np.inf], np.nan).fillna(0)

print("Feature engineering completed!")
print("Number of transactions:", len(features))
print("Number of features:", features.shape[1])

display(features.head())
```

```python
import pandas as pd
import numpy as np

# Copy the expense dataset
ml_data = expenses.copy()

# Create date-based features
ml_data["DayOfWeek"] = ml_data["Date"].dt.dayofweek
ml_data["Month"] = ml_data["Date"].dt.month
ml_data["Hour"] = ml_data["Date"].dt.hour
ml_data["IsWeekend"] = (ml_data["DayOfWeek"] >= 5).astype(int)

# Transform spending amounts
ml_data["LogAmount"] = np.log1p(ml_data["Amount"])

# Compare spending with the category median
category_median = ml_data.groupby("Category")["Amount"].transform("median")

ml_data["CategoryAmountRatio"] = (
    ml_data["Amount"] / category_median.replace(0, np.nan)
).fillna(1)

# Convert text categories into numerical columns
category_features = pd.get_dummies(
    ml_data["Category"], prefix="Category", dtype=int
)

mode_features = pd.get_dummies(
    ml_data["Mode"], prefix="Mode", dtype=int
)

# Combine all features
features = pd.concat(
    [
        ml_data[
            [
                "LogAmount",
                "DayOfWeek",
                "Month",
                "Hour",
                "IsWeekend",
                "CategoryAmountRatio"
            ]
        ],
        category_features,
        mode_features
    ],
    axis=1
)

features = features.replace([np.inf, -np.inf], np.nan).fillna(0)

print("Features created successfully!")
print("Dataset shape:", features.shape)

display(features.head())
```

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)

model.fit(features)

ml_data["Prediction"] = model.predict(features)

ml_data["Result"] = ml_data["Prediction"].map({
    1: "Normal",
    -1: "Unusual"
})

print("Model training completed!")
print(ml_data["Result"].value_counts())
```

```python
# Display unusual transactions
unusual = ml_data[ml_data["Result"] == "Unusual"].sort_values(
    "Amount", ascending=False
)

print("Total transactions analysed:", len(ml_data))
print("Unusual transactions detected:", len(unusual))

display(unusual[["Date", "Category", "Mode", "Amount", "Result"]].head(20))
```
